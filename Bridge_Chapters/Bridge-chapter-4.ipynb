```python
import heapq

def best_first_search_simulation():
    """
    Simulates an agent navigating a state graph using a Value Function.
    Goal: Find path from 'Start' to 'Goal'.
    """
    # 1. The Graph (Adjacency List)
    # Format: Node -> [(Neighbor, Cost)]
    graph = {
        'Start': [('A', 1), ('B', 1)],
        'A': [('C', 1), ('D', 1)],   # Branch A leads to dead ends
        'B': [('E', 1)],             # Branch B leads to Goal
        'E': [('Goal', 1)],
        'C': [], 'D': []             # Dead ends
    }
    
    # 2. The Value Function (Heuristic)
    # Higher score = Closer to goal (Estimated)
    # Notice 'B' has higher value than 'A', guiding the search.
    value_function = {
        'Start': 0.1,
        'A': 0.2, 'B': 0.8,
        'C': 0.0, 'D': 0.0,
        'E': 0.9,
        'Goal': 1.0
    }
    
    # Priority Queue stores tuples: (-Value, Current_Node, Path)
    # We use negative value because heapq is a Min-Heap.
    pq = [(-value_function['Start'], 'Start', ['Start'])]
    
    print("Starting Search...")
    
    while pq:
        # Pop the most promising node (Highest Value)
        neg_val, current, path = heapq.heappop(pq)
        print(f"Exploring: {current} (Value: {-neg_val})")
        
        if current == 'Goal':
            print(f"SUCCESS! Path found: {path}")
            return
            
        # Expand neighbors
        if current in graph:
            for neighbor, cost in graph[current]:
                new_path = path + [neighbor]
                neighbor_val = value_function.get(neighbor, 0)
                heapq.heappush(pq, (-neighbor_val, neighbor, new_path))
                
    print("Failure: Goal not found.")

best_first_search_simulation()
```
