```python
import torch
import torch.nn as nn
import torch.optim as optim

def reinforce_coin_flip():
    """
    Goal: Optimize a parameter 'theta' (probability of heads) 
    to maximize the Reward (getting Heads).
    Constraint: We cannot differentiate through the 'flip' operation.
    """
    # 1. The Parameter (Logits for Heads vs Tails)
    # theta is a learnable weight. 
    theta = torch.tensor([0.0, 0.0], requires_grad=True) 
    optimizer = optim.SGD([theta], lr=0.1)
    
    print("Training a coin to land on Heads (Index 0)...")
    
    for step in range(100):
        optimizer.zero_grad()
        
        # 2. The Policy (Probability Distribution)
        probs = torch.softmax(theta, dim=0)
        
        # 3. The Action (Sampling) - NON-DIFFERENTIABLE STEP
        # We pick an action based on probs
        dist = torch.distributions.Categorical(probs)
        action = dist.sample() 
        
        # 4. The Reward
        # Reward is 1.0 if Heads (action 0), else 0.0
        reward = 1.0 if action.item() == 0 else 0.0
        
        # 5. The Loss (Negative Objective) using Log-Derivative Trick
        # Loss = - (Reward * log_prob(action))
        # We assume baseline b=0 for simplicity
        log_prob = dist.log_prob(action)
        loss = -(reward * log_prob)
        
        loss.backward()
        optimizer.step()
        
        if step % 20 == 0:
            print(f"Step {step}: Prob(Heads) = {probs[0].item():.4f}")

    print(f"Final Prob(Heads): {torch.softmax(theta, dim=0)[0].item():.4f}")

reinforce_coin_flip()
```
