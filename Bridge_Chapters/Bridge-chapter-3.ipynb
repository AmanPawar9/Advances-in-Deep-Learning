```python
import torch
import numpy as np

def adaptive_compute_simulation():
    """
    Simulates a system that allocates more 'search budget' 
    to uncertain predictions (High Entropy).
    """
    # 1. Simulate Model Predictions (Softmax probabilities)
    # Case A: Confident Model (Low Entropy)
    probs_confident = torch.tensor([0.9, 0.05, 0.05])
    
    # Case B: Confused Model (High Entropy)
    probs_confused = torch.tensor([0.34, 0.33, 0.33])
    
    def calculate_entropy(probs):
        # H = - sum(p * log(p))
        return -torch.sum(probs * torch.log(probs + 1e-9))
    
    h_conf = calculate_entropy(probs_confident)
    h_unsure = calculate_entropy(probs_confused)
    
    print(f"Entropy (Confident): {h_conf:.4f}")
    print(f"Entropy (Confused):  {h_unsure:.4f}")
    
    # 2. Adaptive Branching Logic (Chapter 3)
    # Formula: k = 1 + lambda * Entropy
    lambda_param = 5.0
    
    k_conf = int(1 + lambda_param * h_conf)
    k_unsure = int(1 + lambda_param * h_unsure)
    
    print("\n--- Adaptive Branching (Compute Allocation) ---")
    print(f"Confident Model gets {k_conf} search branches.")
    print(f"Confused Model gets  {k_unsure} search branches.")
    print("Interpretation: We spend compute only where we are unsure.")

adaptive_compute_simulation()
```
