```python
import numpy as np

def memory_comparison(L=1000):
    """
    Demonstrates the memory difference between storing the Full Attention Matrix
    vs. storing the Low-Rank factors (Key/Query).
    """
    
    # 1. The Low-Rank Factors (The "Hidden State" view)
    # Vectors u and v of length L
    u = np.random.randn(L, 1) # e.g., The "Query" or State Output
    v = np.random.randn(L, 1) # e.g., The "Key" or State Input
    
    memory_factors = u.nbytes + v.nbytes
    
    # 2. The Full Interaction Matrix (The "Attention" view)
    # M = u * v.T (Outer Product)
    M = u @ v.T 
    
    memory_matrix = M.nbytes
    
    print(f"Sequence Length: {L}")
    print(f"Memory (Factors): {memory_factors / 1024:.2f} KB")
    print(f"Memory (Matrix):  {memory_matrix / 1024:.2f} KB")
    print(f"Compression Ratio: {memory_matrix / memory_factors:.1f}x")
    
    # Verification of Rank
    rank = np.linalg.matrix_rank(M)
    print(f"Rank of Matrix: {rank}")

memory_comparison(L=2048)
```
